## 使用subprocess


```python
import subprocess

result = subprocess.run(['echo', 'Hello, World!'], capture_output=True, text=True,shell = True)

print("返回码:", result.returncode) # 0 表示成功
print("标准输出:\n", result.stdout)
print("标准错误:", result.stderr)

result = subprocess.run(['ipconfig','-all'],capture_output=True,text=True,shell =True)

print("返回码:", result.returncode) # 0 表示成功
print("标准输出:\n", result.stdout)
print("标准错误:", result.stderr)
```
